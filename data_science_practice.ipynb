```python
name = input("what is your name?")
salary = float(input("what is your salary?"))
commission_rate = float(input("what is your commission rate (as a decimal)"))
bonus = float(input("what is your bonus?"))
bonus_threshold = 500
commission_rate_threshold = 0.15

if bonus <= bonus_threshold and commission_rate <= commission_rate_threshold:
    total_compensation = salary + bonus + (salary * commission_rate)
    print(f"total compensation for {name} is: ${total_compensation:.2f}")
else:
    print(f"{name} is not eligible for a bonus")
```

```python
def grade(value):
    if value >= 75:
        return "Grade A"
    elif value >= 65:
        return "Grade B"
    elif value >= 55:
        return "Grade C"
    elif value >= 45:
        return "Grade D"
    elif value >= 40:
        return "Grade E"
    else:
        return "Grade F"

def main():
    value = int(input("Enter your Score: "))
    print(grade(value))

# To run the code, you can now call main()
main()
```

```python
def main():
    target = 2000000
    weekly_goal = 50000
    weeks = weeks_to_target(target, weekly_goal)
    print(f"Israel hits his target in {weeks} weeks")

def weeks_to_target(target, weekly_goal):
    saved = 0
    weeks = 0
    while saved < target:
        saved += weekly_goal
        weeks += 1
        print(f"week {weeks}: $ {saved:,}")

    return weeks

main()
```
